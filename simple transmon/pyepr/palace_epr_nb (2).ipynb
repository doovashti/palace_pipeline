# Palace -> pyEPR analysis

Takes Palace eigenmode output and runs it through pyEPR's real
`QuantumAnalysis.analyze_variation()` -- the identical Hamiltonian code path
`EPR_master_nb.ipynb` uses for Ansys.

**What gets substituted:** pyEPR's `ProjectInfo` and `DistributedAnalysis`
(the Ansys-specific stages) are replaced by reading Palace's CSVs.
`QuantumAnalysis` is kept unmodified, so every formula -- `chi_O1`, `f_1`,
`chi_ND`, `f_ND`, `ZPF` -- is computed by pyEPR itself.

| pyEPR needs | Palace file |
|---|---|
| mode frequencies, Q | `eig.csv` |
| junction participations $p_{mj}$ | `port-EPR.csv` |

Requires `palace_epr.py` in the same folder and the `qiskit_metal` env.

````python
%load_ext autoreload
%autoreload 2

import os
import numpy as np
import pandas as pd
from IPython.display import display

from palace_epr import (
    load_palace, validate_junction_ports, identify_modes, analyze,
    print_fancy_results, compare_with_ansys, get_g,
)
````

## 1. Configuration

**AMR runs** write one folder per refinement pass (`iteration1`,
`iteration2`, ...). Point at the **highest-numbered** one -- that is the most
converged mesh.

**`MODES`** selects which modes enter the Hamiltonian analysis. This is
required: the numerical diagonalization builds a `fock_trunc ** n_modes`
Hilbert space, so including every mode Palace found will exhaust memory.
The Ansys notebook does the same subsetting via `modes_to_analyse`.

````python
POSTPRO = 'iteration3'    # folder with eig.csv + port-EPR.csv
PALACE_CONFIG = 'postpro_1/palace_config_lowmodes_save_4_resolved.json'
# Saved input JSON: it must contain the inductive LumpedPort spanning the JJ gap.

LJ = [10e-9]      # junction inductance(s), Henries  -- must match Palace config "L"
CJ = [3e-15]      # junction capacitance(s), Farads  -- only feeds Ec, not chi

MODES      = [0, 1, 2, 3]   # 0-based Palace mode indices to analyze
MODE_NAMES = None           # labels for the SELECTED modes, or None
QUBIT_IND  = None           # position within MODES; None -> auto

COS_TRUNC, FOCK_TRUNC = 6, 7
````

## 2. Load Palace results

````python
eig, Pmj = load_palace(POSTPRO)
jj_ports = validate_junction_ports(POSTPRO, LJ, CJ, PALACE_CONFIG)

print(f'{len(eig)} modes loaded from {POSTPRO}\n')
display(eig.style.set_caption('eig.csv').format(
    {'f_GHz': '{:.6f}', 'fim_GHz': '{:.4e}', 'Q': '{:.4e}'}))
display(pd.DataFrame(Pmj, columns=[f'p[{j+1}]' for j in range(Pmj.shape[1])])
        .style.set_caption('port-EPR.csv  (signed junction participations)')
        .format('{:.6e}'))
````

## 3. Identify the modes

The qubit is the mode with $|p| \approx 1$ -- nearly all of its inductive
energy sits in the junction.

If $|p|$ is **split across two nearly degenerate modes**, they are hybridised:
the qubit is shared between them. Per-mode quantities then depend on
arbitrary mode ordering, and you should compare **sums over the pair**
instead of individual values.

````python
tbl = identify_modes(eig, Pmj, fock_trunc=FOCK_TRUNC)
````

## 4. Run the analysis

This writes a pickle matching `QuantumAnalysis.__init__`'s schema, loads it,
and calls the real `analyze_variation`.

Participation renormalisation is disabled (`renorm_pj=False`): it requires
Ansys field-energy integrals Palace does not produce, and Palace's
participations are already normalised. `analyze()` asserts `_Pm_norm == 1`
to confirm no rescaling was applied.

````python
res, epra = analyze(
    eig, Pmj,
    Ljs=LJ, Cjs=CJ,
    modes=MODES,
    cos_trunc=COS_TRUNC, fock_trunc=FOCK_TRUNC,
    mode_names=MODE_NAMES,
)

M = len(res['f_0'])
names = MODE_NAMES or [f'Mode {i+1}' for i in sorted(MODES)]

print(f'analyzed Palace mode indices: {sorted(MODES)}')
print(f'returned {M} modes; keys: {list(res.keys())[:8]} ...')
````

## 5. Results

Same tables as `Master.print_fancy_results()`:

* **Mode frequencies** -- linear (bare EM solve), dressed (1st-order PT),
  numerically diagonalised
* **g** -- couplings from the qubit to every other mode
* **kappa / T_1** -- from Palace's Q factors
* **chi** -- analytic and numerical. Diagonal is anharmonicity, off-diagonal
  is cross-Kerr
* **Q factors** and **participations**

````python
qubit_ind = print_fancy_results(res, names=names,
                                qubit_ind=QUBIT_IND, display_fn=display)
print(f'\nqubit mode: {names[qubit_ind]}  (position {qubit_ind})')
````

**Compare like with like.** `chi_O1` (analytic, 1st-order PT) and `chi_ND`
(numerical diagonalisation) are different quantities and diverge for
hybridised modes. Never compare one against the other.

**Match modes by physics, not index.** Palace sorts by frequency; HFSS uses
its own ordering. Use the participation table to pair them up.

````python
ansys_f_GHz = [4.07077202242, 4.11526568004, 9.65822006337, 10.9729659308]
ansys_Q     = [3388236.8379831347, 1581847.2987456485, 1572.4979101605127, 5.98989517679776]
ansys_chi_ND = [[109.61908412887192, -106.29673617440127, 0.03701157488250732, 0.001026961856842041], [-106.29673617440127, 128.14044630638216, 1.2849775446414946, 0.034299837907791134], [0.03701157488250732, 1.2849775446414946, 0.0031497618179321288, 0.00011450469207763671], [0.001026961856842041, 0.034299837907791134, 0.00011450469207763671, 1.2122344970703125e-06]]
ansys_anharm = [109.61908412887192, 128.14044630638216, 0.0031497618179321288, 1.2122344970703125e-06]
ansys_p      = [[0.1964837381233744], [0.7776849795475617], [0.002128411801473461], [2.9693247139557485e-05]]


compare_with_ansys(
    res,
    ansys_f_GHz=ansys_f_GHz,
    ansys_Q=ansys_Q,
    ansys_chi_O1=ansys_chi_O1,
    ansys_chi_ND=ansys_chi_ND,
    ansys_p=ansys_p,
    names=names,
    pair=(0, 1),          # the hybridised pair, for the sum-rule test
    display_fn=display,
)
````

### Interpreting the sum rules

If **$\sum p$ agrees** between tools but **$\sum p^2$ differs**, both solvers
found the same total junction coupling but distributed it differently across
the hybridised pair.

Since anharmonicity scales as $p^2$, a modest redistribution of participation
is amplified into a large disparity in anharmonicity -- and, through the
mode's coupling to the lossy port, in $Q$ as well.

In that case there is **one** underlying discrepancy (how the near-degenerate
pair splits), not several. What sets the splitting is the geometry and the
coupling between those two modes, so that is what to investigate -- starting
with whether both tools are running the same geometry revision.

## 7. Optional: O1 vs ND divergence

`chi_O1` is first-order perturbation theory; `chi_ND` is exact
diagonalisation. They agree when modes are well separated and weakly coupled,
and diverge when they are not.

A large gap on specific modes is a quantitative signal that perturbation
theory is breaking down there -- independent evidence of hybridisation.

````python
o1 = np.diag(res['chi_O1'].to_numpy())
nd = np.diag(res['chi_ND'].to_numpy())

df = pd.DataFrame({'chi_O1 (MHz)': o1, 'chi_ND (MHz)': nd}, index=names)
df['|O1 - ND| (MHz)'] = np.abs(o1 - nd)
df['rel. gap (%)'] = 100 * np.abs(o1 - nd) / np.abs(nd)
display(df.style.set_caption(
    'Perturbative vs exact anharmonicity -- large gap = PT breaking down')
    .format('{:.4f}'))
````
